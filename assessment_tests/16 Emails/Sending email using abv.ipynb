```python
import smtplib
import getpass
import ssl
```

```python
smtp_object = smtplib.SMTP_SSL("smtp.abv.bg",465) # ABV needs SSL.
```

```python
email = getpass.getpass("Please enter email")
password = getpass.getpass("Please enter password")
```

```python
smtp_object.ehlo()
```

```python
smtp_object.login(email,password)
```

```python
from_email = email
to_email = email
subject = input("Please input subject of the email: ")
message = input("Please input msg of the email: ")

msg = f"From: {from_email}\nTo: {to_email}\nSubject: {subject}\n\n{message}" # the old way in abv is deprecated so we have to add From: and To:

smtp_object.sendmail(from_email,to_email,msg)
```

```python
smtp_object.quit()
```
