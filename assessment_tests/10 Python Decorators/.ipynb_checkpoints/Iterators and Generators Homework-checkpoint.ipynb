# Iterators and Generators Homework 

### Problem 1

Create a generator that generates the squares of numbers up to some number N.

```python
def gensquares(N):
    for x in range(N):
        yield x**2
```

```python
for x in gensquares(10):
    print(x)
```

### Problem 2

Create a generator that yields "n" random numbers between a low and high number (that are inputs). <br>Note: Use the random library. For example:

```python
import random

random.randint(1,10)
```

```python
def rand_num(low,high,n):
    for _ in range(n):
        yield random.randint(1,10)
```

```python
for num in rand_num(1,10,12):
    print(num)
```

### Extra Credit!
Can you explain what *gencomp* is in the code below? (Note: We never covered this in lecture! You will have to do some Googling/Stack Overflowing!) <br><br> ANSWER: Its a generator comprehenstion which is like a list comprehension but instead of saving everything to memory it just returns the next object so here first it will return 4 when we call next(gencomp) and then it will return 5.

```python
my_list = [1,2,3,4,5]

gencomp = (item for item in my_list if item > 3)

for item in gencomp:
    print(item)
```
