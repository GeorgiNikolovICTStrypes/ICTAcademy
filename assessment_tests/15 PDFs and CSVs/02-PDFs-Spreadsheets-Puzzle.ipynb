## Task Two: Download the PDF from the Google Drive link and find the phone number that is in the document. 

```python
# You should get this phone number
# 505 503 4455

import PyPDF2
import re
regex_pattern = re.compile(r"\d{3}.\d{3}.\d{4}")

with open('Find_the_Phone_Number.pdf', 'rb') as pdf_file:

    pdf_reader = PyPDF2.PdfReader(pdf_file)

    num_pages = len(pdf_reader.pages)

    for i in range(num_pages):
        curr_page = pdf_reader.pages[i]
        page_text = curr_page.extract_text()

        result = regex_pattern.search(page_text)
        if result:
            print(f"Found the phone number {result.group(0)} on page {i}")
```
