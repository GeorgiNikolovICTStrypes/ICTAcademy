```python
import zipfile
import os
import re
```

```python
zip_obj = zipfile.ZipFile('unzip_me_for_instructions.zip','r')
```

```python
zip_obj.extractall('unzip_me_for_instructions')
```

```python
# get the directory we want to work in
filepath = 'C:\\Users\\GeorgiNikolov\\Desktop\\Academy\\Python Repo\\ICTAcademy\\assessment_tests\\11 advanced modules\\unzip_me_for_instructions\\extracted_content'
# Construct the regex
regex = re.compile("(\d\d\d)-\d\d\d-\d\d\d\d")
matches = []
# Explore it so for each directory we get 2 lists of subdirs and files
# So for each dir we explore each file inside it and we glue dir + file to get the path to that file.
# After that we explore each files contents by opening it with readmode with a context manager
# Then we explore each line and we apply our regex to it
# if we fidn a match we add it to our matches list.
for dirpath, sub_foulder, files in os.walk(filepath):
    
    for f in files:
        if f == "Instructions-checkpoint.txt":
            continue
        directory = os.path.join(dirpath, f)
        # print(f"This is a file: {f}")
        
        with open(directory,"r") as file_desc:
            for num, line in enumerate(file_desc):
                match = regex.search(line)
                if match:
                    matches.append({"file" : directory, "line": num, match: match.group()})
    print('\n')

print(matches)
```
